# 介紹
使用pyutogui來完成小恐龍遊戲的機器人
小恐龍是讚chrome中使用 `chrome://dino`中啟動

邏輯:
- 不斷監視畫面
- 讀取某個位置的顏色
- 若此位置上出現(83,83,83)的顏色
- 跳起來

```python
import pyautogui as pyg
pyg.sleep(2) # sleep 2 sec.
pyg.press('space')#啟動 
#無窮迴圏 
while True:
  im = pyg.screenshot()
  obs = False#假設沒有仙人掌
  for i in range(15):
    c = im.getpixel((1277+i,313))
    if c==(83,83,83):
        break
    if obs == True:   
       
     pyg.press('space')
```
