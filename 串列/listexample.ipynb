# 簡介
串列是一個儲存多個元素 (整數 浮點數 字串)的容器
串列中每個元素都是用逗號隔開 左右有中括號[]

```python
#score 是一個串列 裡面每個元素都是整數
score=[78,29,56,80,99]
print(score)
```

## 索引值
我們可以使用索引值來存取串列中的元素
注意:索引值由0開始 

```python
print(score[0])#印出第一個元素
print(score[1])#印出第二個元素
```

## 串列內容的個數
len(a)取得a個個串列中元素的個數
len是lenth(長度)的意思

```python
print(len(score))
print(score[4])#取最後一個元素
print(score[5-1])#取最後一個元素
print(score[len(score)-1])#取最後一個元素
print(score[-1])#取最後一個元素
```

## 新增元素
使用`append()` 新增元素到串列的尾端

```python
print(score)
score.append(78)
score.append(93)
print(score)
```

```python
print(score)
score[2]=70#將70指定給索引值為2的元素 把地3項改成70
print(score)
```

```python
weight=[12]*6
print(weight)
```

## 走訪串列的三種寫法


```python
print(score)
for i in range(len(score)):
    print(score[i])
```

```python
#對score中的每個元素 一次取一個 放在e中
for e in score:
    print(e)
```

```python
for i ,e in enumerate(score):
    print(i,e)
```

## 計算費伯納西數列的第91個元素
一個費伯納西數列如下:
1,1,2,3,5,8.....
請問第91個元素為何

提示:
另$x$是一個費伯納數列,$x[i]$ 是第$i$個數字
我們會得到$x[n]=x[n-1]=x[n+2]$

```python
# 初始化串列 包含91個0
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
i=2
fib[i]=fib[i-2]+fib[i-1]
print(fib)
i=3
fib[i]=fib[i-2]+fib[i-1]
i=4
fib[i]=fib[i-2]+fib[i-1]
print(fib)
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]

print(fib[90])
```
