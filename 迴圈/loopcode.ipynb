# Loop
Loop 指令可以讓程式碼依我的條件不斷地執行,直到條件滿足為止。

```python
#底下程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list (range(10))
```

```python
for i in range(10):
    print(i+1)
```

# 計算1+2+3+....+10

```python
sum=0
for i in range(10):
    sum=sum+(1+i)
print(sum)

```

印出底下圖  
&  
&&  
&&&  
&&&&  
&&&&&  

```python
for i in range(5):
    print("&"*(i+1))
```

印出   
1/*1=1  
1/*2=2  
1/*3=3  
...  
1/*9=9  

```python
print(f"1*1={1*1}")
print(f"1*2={1*2}")
```

```python
for i in range(10):
    print(f"1*{(i+1)}={1*(i+1)}")
```

# Range 

python:
- 懶惰是種美德
- 由0開頭
- 包含頭,不包含尾


```python
print(list(range(5)))
#區間,由0開始,到5-1結束
print(list(range(0,5)))
#區間,由1開始,到5-1結束
print(list(range(1,5)))
#區間,由1開始,到5-1結束,每次跳2個
print(list(range(1,5,2)))
```

```python
print(list(range(2,101,2)))
```
